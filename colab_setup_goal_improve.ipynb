## Step 3 — Find the Correct Checkpoint Path

```python
import glob

# Auto-discover the checkpoint .pt file inside action_sampling_weights_002
checkpoints = glob.glob('/content/weights/**/model*.pt', recursive=True)
checkpoints += glob.glob('/content/weights/**/*.pt', recursive=True)
checkpoints += glob.glob('/content/weights/**/*.pth', recursive=True)

print('Found checkpoint files:')
for c in checkpoints:
    print(f'  {c}')

# Set the checkpoint path
CHECKPOINT_PATH = checkpoints[0] if checkpoints else None
print(f'\nUsing: {CHECKPOINT_PATH}')
```

## Step 6 — Run Goal Detection

```python
import sys
sys.path.insert(0, '/content/Football_')
sys.path.insert(0, '/content/ball-action-spotting')

from src.soccernet_goal_detector import SoccerNetGoalDetector

detector = SoccerNetGoalDetector(
    checkpoint_path=CHECKPOINT_PATH,
    repo_path='/content/ball-action-spotting',
    device='cuda',
    gaussian_sigma=3.0,
    peak_min_height=0.2,
    peak_min_distance_frames=15,
    replay_merge_seconds=45.0,
    batch_size=8,
)

goals = detector.detect(VIDEO_PATH)

print('\n===== GOAL DETECTION RESULTS =====')
if goals:
    for i, (ts, conf) in enumerate(goals, 1):
        m, s = int(ts // 60), int(ts % 60)
        print(f'  Goal #{i}: {m:02d}:{s:02d} ({ts:.1f}s) — confidence: {conf:.3f}')
else:
    print('  No goals detected.')
```

## Step 7 — (Optional) Visualize Goal Confidence Curve

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.ndimage import gaussian_filter1d

# Re-run internal steps to plot the curve
frames, source_fps, skip_n = detector._preprocess_video(VIDEO_PATH)
sequences = detector._build_input_tensors(frames)
all_scores = detector._run_inference(sequences)

from src.soccernet_goal_detector import GOAL_CLASS_INDEX, EFFECTIVE_FPS
goal_scores = all_scores[:, GOAL_CLASS_INDEX]
smoothed = gaussian_filter1d(goal_scores, sigma=3.0)

actual_fps = source_fps / skip_n
timestamps = np.arange(len(goal_scores)) / actual_fps

plt.figure(figsize=(16, 4))
plt.plot(timestamps, goal_scores, alpha=0.4, color='blue', label='Raw Goal Score')
plt.plot(timestamps, smoothed, color='red', linewidth=2, label='Smoothed (sigma=3)')
plt.axhline(0.2, color='green', linestyle='--', label='Min Peak Height (0.2)')
for ts, conf in goals:
    plt.axvline(ts, color='gold', linewidth=2, label=f'Goal @ {ts:.1f}s')
plt.xlabel('Time (seconds)')
plt.ylabel('Goal Confidence')
plt.title('SoccerNet Goal Confidence Curve')
plt.legend()
plt.tight_layout()
plt.savefig('/content/goal_confidence_curve.png', dpi=150)
plt.show()
print('Saved: /content/goal_confidence_curve.png')
```
